### **PROMPT 1 : PENGOLAHAN DATA SEISMIK REFRAKSI**



#### ***Struktur Prompt***

**Judul: Analisis Data Seismik Refraksi untuk Identifikasi Litologi dan Struktur Bawah Permukaan**

Bertindaklah sebagai ahli geofisika terapan dan dosen pembimbing praktikum yang memiliki keahlian dalam metode seismik refraksi, analisis gelombang P, pemodelan kecepatan, dan interpretasi geologi. Lakukan pengolahan data seismik refraksi berdasarkan hubungan antara waktu tiba pertama gelombang P dan jarak sumber-geofon. Gunakan pendekatan **Time-Distance Curve** dan **Intercept Time** untuk memperkirakan kecepatan serta kedalaman lapisan bawah permukaan. Sajikan proses pengolahan secara runtut, lengkap dengan persamaan, contoh perhitungan, kode Python, visualisasi, hasil pemodelan, dan interpretasi geologi.

#### ***R — Role (Peran)***

Berperan sebagai ahli geofisika terapan dan dosen pembimbing praktikum Geofisika yang memahami metode seismik refraksi, first arrival picking, analisis kurva waktu-jarak, Hukum Snellius, pemodelan kecepatan bawah permukaan, serta interpretasi litologi berdasarkan kecepatan gelombang P.

#### ***T — Task (Tugas)***

Lakukan pengolahan dan analisis data seismik refraksi untuk:
1. Mengidentifikasi waktu tiba pertama gelombang P (*first arrival time*).
2. Menyusun tabel hubungan jarak geofon dengan waktu tempuh gelombang.
3. Membuat **Time-Distance Curve** dan mengidentifikasi segmen garis yang mewakili lapisan berbeda.
4. Menghitung kecepatan gelombang P setiap lapisan dari kemiringan kurva.
5. Mengestimasi kedalaman dan ketebalan lapisan menggunakan metode **Intercept Time**.
6. Menentukan urutan lapisan berdasarkan hubungan kecepatannya.
7. Membuat model sederhana bawah permukaan berupa penampang lapisan.
8. Menginterpretasikan kemungkinan material/litologi berdasarkan nilai kecepatan dan konteks geologi.
9. Menjelaskan keterbatasan model apabila data hanya mendukung interpretasi 1D atau 2D sederhana.

#### ***C — Context (Konteks)***

Metode seismik refraksi memanfaatkan perbedaan kecepatan rambat gelombang seismik pada material bawah permukaan. Gelombang yang mengalami refraksi kritis pada batas lapisan dapat menghasilkan gelombang kepala yang direkam pada geofon dan digunakan untuk memperkirakan kecepatan lapisan.

Analisis ditujukan untuk membantu mengenali lapisan tanah, sedimen, material lapuk, hingga batuan dasar (*bedrock*). Interpretasi tidak boleh hanya didasarkan pada angka kecepatan, tetapi juga harus mempertimbangkan kondisi geologi dan keterbatasan data.

#### ***I — Input (Masukan)***

Data yang digunakan meliputi:
- Waktu tiba pertama gelombang P pada setiap geofon.
- Jarak antar-geofon.
- Posisi sumber/shot point.
- Jumlah geofon dan panjang lintasan.
- Arah dan posisi lintasan jika tersedia.
- Data pengukuran dalam format CSV, Excel, atau tabel.
- Informasi geologi permukaan jika tersedia.

Jika data lapangan belum tersedia, gunakan **data simulasi** yang diberi label secara eksplisit sebagai data simulasi. Jangan menyatakan data simulasi sebagai data hasil pengukuran lapangan.

#### ***K — Constraints (Batasan)***

1. Gunakan konsep **Time-Distance Curve** dan **Intercept Time** sesuai kebutuhan data.
2. Tampilkan persamaan dan definisi setiap parameter sebelum melakukan perhitungan.
3. Pertahankan satuan jarak dalam meter dan waktu dalam sekon atau milisekon dengan konversi yang jelas.
4. Gunakan prinsip perambatan gelombang dan Hukum Snellius secara konsisten.
5. Jangan mengarang nilai pengukuran yang tidak tersedia.
6. Jika menggunakan data simulasi, jelaskan bahwa data tersebut hanya untuk demonstrasi.
7. Jangan menyimpulkan jenis batuan secara pasti hanya berdasarkan satu nilai kecepatan.
8. Gunakan bahasa ilmiah tingkat mahasiswa S1 Geofisika.
9. Jika kode Python digunakan, kode harus dapat dijalankan dengan library yang umum seperti NumPy, Pandas, dan Matplotlib.

#### ***O — Output (Keluaran)***

Hasil analisis harus memuat:
- Penjelasan singkat prinsip metode.
- Tabel data awal.
- Tabel hasil pengolahan.
- Persamaan yang digunakan.
- Contoh perhitungan kecepatan.
- Perhitungan kedalaman/ketebalan lapisan.
- Grafik **Time-Distance Curve**.
- Model atau skema lapisan bawah permukaan.
- Kode Python apabila diperlukan.
- Interpretasi setiap lapisan.
- Pembahasan keterbatasan dan ketidakpastian hasil.

#### ***Q — Quality Criteria (Kriteria Kualitas)***

Output dianggap layak apabila:
1. Seluruh satuan konsisten.
2. Kemiringan kurva dan kecepatan mempunyai hubungan yang benar.
3. Kedalaman lapisan dihitung menggunakan persamaan yang sesuai.
4. Grafik mempunyai label sumbu, legenda, dan satuan.
5. Model bawah permukaan konsisten dengan hasil perhitungan.
6. Interpretasi geologi dinyatakan sebagai interpretasi, bukan fakta mutlak.
7. Tidak terdapat angka yang dibuat tanpa penjelasan sumbernya.

#### ***V — Verification (Verifikasi)***

Sebelum memberikan hasil akhir, lakukan pemeriksaan terhadap:
- konsistensi satuan waktu dan jarak;
- hasil konversi milisekon ke sekon jika diperlukan;
- kemiringan setiap segmen kurva;
- kecepatan setiap lapisan;
- kedalaman dan ketebalan lapisan;
- kecocokan model dengan data;
- serta konsistensi interpretasi dengan hasil geofisika.



#### ***Struktur Prompt***

**Judul: Analisis Data Geolistrik Resistivitas untuk Identifikasi Litologi dan Akuifer Bawah Permukaan**

Lakukan pengolahan dan analisis untuk:
1. Mengidentifikasi konfigurasi elektroda yang digunakan.

3. Menghitung resistivitas semu menggunakan data arus dan beda potensial.
4. Menyusun tabel hasil perhitungan.

8. Menginterpretasikan kemungkinan litologi berdasarkan rentang resistivitas dan informasi geologi.
9. Mengidentifikasi zona yang secara geofisika berpotensi mengandung air tanah.
10. Menjelaskan ambiguitas interpretasi karena satu nilai resistivitas dapat mewakili lebih dari satu kondisi geologi.

#### ***C — Context (Konteks)***

Metode geolistrik resistivitas menentukan variasi sifat listrik bawah permukaan dengan menginjeksikan arus dan mengukur beda potensial. Nilai resistivitas dipengaruhi oleh mineral penyusun batuan, porositas, tingkat kejenuhan, kandungan air, salinitas, dan kondisi struktur batuan.

Dalam kajian hidrogeofisika, zona dengan resistivitas tertentu dapat menjadi indikasi keberadaan air tanah. Namun, identifikasi akuifer harus mempertimbangkan konteks geologi dan tidak boleh dilakukan hanya berdasarkan satu nilai resistivitas.

#### ***I — Input (Masukan)***

Data dapat berupa:
- kuat arus listrik (**I**);
- beda potensial (**ΔV**);
- jarak elektroda arus dan elektroda potensial;
- konfigurasi Schlumberger atau Wenner;
- nilai AB, MN, AB/2, atau parameter geometrik lainnya;
- koordinat titik pengukuran;
- data CSV atau Excel;
- informasi litologi/geologi regional;
- data sumur atau informasi hidrogeologi jika tersedia.

Jika data lapangan belum tersedia, gunakan data simulasi dan beri label dengan jelas.

#### ***K — Constraints (Batasan)***

1. Gunakan faktor geometri sesuai konfigurasi elektroda.
2. Tampilkan persamaan **ρa = K ΔV/I** beserta definisi setiap parameter.
3. Jangan mencampurkan rumus konfigurasi Wenner dan Schlumberger.
4. Gunakan satuan SI secara konsisten.
5. Jika melakukan inversi, jelaskan jenis model dan asumsi yang digunakan.
6. Jangan menganggap resistivitas rendah selalu berarti air tanah karena dapat pula disebabkan oleh lempung atau material konduktif lain.
7. Jangan membuat data lapangan fiktif tanpa label.
8. Gunakan Python dan library yang relevan apabila diminta menghasilkan kode.

#### ***O — Output (Keluaran)***

Berikan:
- penjelasan prinsip metode;
- identifikasi konfigurasi;
- tabel faktor geometri;
- tabel resistivitas semu;
- contoh perhitungan;
- grafik data;
- hasil inversi dan parameter model jika tersedia;
- kode Python;
- penampang/model resistivitas;
- interpretasi litologi;
- identifikasi zona prospek akuifer beserta alasan geofisikanya;
- pembahasan ketidakpastian.

#### ***Q — Quality Criteria (Kriteria Kualitas)***

Hasil dianggap memenuhi kriteria apabila:
1. Faktor geometri sesuai konfigurasi.
2. Resistivitas semu dihitung tanpa kesalahan satuan.
3. Grafik menggunakan skala dan label yang sesuai.
4. Hasil inversi dijelaskan bersama parameter modelnya.
5. Interpretasi tidak melampaui informasi yang diberikan.
6. Zona akuifer dijelaskan berdasarkan kombinasi resistivitas dan konteks geologi.

#### ***V — Verification (Verifikasi)***

Periksa ulang konfigurasi elektroda, faktor geometri, nilai arus dan beda potensial, resistivitas semu, satuan, hasil inversi, serta hubungan model dengan kondisi geologi.

Berperan sebagai ahli geofisika terapan dan dosen pembimbing yang memahami teori elektromagnetik, akuisisi data MT, tensor impedansi, apparent resistivity, phase, dimensionality analysis, inversi 1D/2D, dan interpretasi sistem panas bumi.

Lakukan pengolahan dan analisis untuk:
1. Memeriksa kualitas data medan listrik dan medan magnet.
2. Mengidentifikasi komponen Ex, Ey, Hx, Hy, dan Hz jika tersedia.

8. Membuat penampang distribusi resistivitas bawah permukaan.

Dalam eksplorasi panas bumi, zona konduktif dapat berkaitan dengan mineral lempung hasil alterasi hidrotermal, fluida, atau material lain yang konduktif. Zona resistif dapat memiliki beberapa kemungkinan interpretasi. - medan magnet vertikal Hz jika tersedia;
- frekuensi atau periode;
- impedansi elektromagnetik;
- resistivitas semu dan fase jika sudah dihitung;
- koordinat stasiun MT;
- orientasi sensor;
- data geologi dan geokimia;
- data topografi;
- informasi sumur atau manifestasi panas bumi jika tersedia.

Jika data mentah tidak tersedia, gunakan data simulasi atau data contoh dan beri label secara eksplisit.

#### ***K — Constraints (Batasan)***

1. Jelaskan hubungan antara impedansi, resistivitas semu, fase, dan frekuensi/periode.
2. Gunakan persamaan MT yang sesuai dan definisikan setiap variabel.
3. Pertahankan satuan dan konvensi koordinat secara konsisten.
4. Jangan menganggap setiap zona konduktif sebagai reservoir panas bumi.
5. Jika melakukan inversi, jelaskan parameter model, regularisasi, dan ukuran kecocokan data bila tersedia.
6. Bedakan antara hasil observasi, hasil inversi, dan interpretasi geologi.
7. Data simulasi tidak boleh dinyatakan sebagai hasil pengukuran lapangan.
8. Jika menggunakan kode Python, gunakan library yang sesuai dan jelaskan fungsi setiap bagian kode.
9. Jangan menghilangkan data noisy tanpa menjelaskan kriteria atau dasar penyaringannya.

#### ***O — Output (Keluaran)***

Hasil harus memuat:
- penjelasan prinsip MT;
- pemeriksaan data;
- persamaan impedansi dan resistivitas semu;
- tabel hasil;
- kurva resistivitas semu;
- kurva fase;
- kode Python jika relevan;
- hasil inversi 1D/2D;
- penampang resistivitas;
- identifikasi zona konduktif dan resistif;
- interpretasi geologi dan panas bumi;
- keterbatasan serta alternatif interpretasi.

#### ***Q — Quality Criteria (Kriteria Kualitas)***

Output dinilai layak apabila:
1. Hubungan frekuensi/periode dan resistivitas dijelaskan dengan benar.
2. Satuan dan konvensi data konsisten.
3. Kurva resistivitas dan fase diberi label yang lengkap.
4. Hasil inversi disertai penjelasan mengenai model dan kecocokan data.
5. Interpretasi zona konduktif/resistif mempertimbangkan lebih dari satu kemungkinan.
6. Kesimpulan tidak melampaui informasi yang tersedia.

#### ***V — Verification (Verifikasi)***

Lakukan pemeriksaan akhir terhadap:
- komponen medan listrik dan magnet;
- satuan frekuensi/periode;
- impedansi;
- resistivitas semu;
- fase;
- kualitas kurva;
- hasil inversi dan kecocokan model;
- penampang resistivitas;
- serta konsistensi interpretasi dengan data geologi/geokimia pendukung.

Jika terdapat data yang tidak cukup untuk melakukan salah satu tahap, nyatakan keterbatasannya dan jangan membuat nilai pengganti tanpa label.
